```python
import numpy as np

montos = np.array([
    37.00, 25.00, 10.00, 5.00, 12.25,
    20.00, 8.00, 1.00, 15.00, 2.50
])

media = np.mean(montos)
mediana = np.median(montos)
desviacion = np.std(montos)

percentil_25 = np.percentile(montos, 25)
percentil_50 = np.percentile(montos, 50)
percentil_75 = np.percentile(montos, 75)

print("Media:", media)
print("Mediana:", mediana)
print("Desviación estándar:", desviacion)
print("Percentil 25:", percentil_25)
print("Percentil 50:", percentil_50)
print("Percentil 75:", percentil_75)
```
